## Join data frames keeping `DATE` and `PRCP` columns

```python
df_seattle.head(2)
```

```python
df_portland.head(2)
```

**What type of join should we do?**

An **outer** join keeps every date that appears in either data set. A date missing from one city then shows up as `NaN` instead of being silently dropped, which an inner join would do.

##### Use a join to keep all dates present

```python
df = pd.merge(df_seattle[['DATE', 'PRCP']],
              df_portland[['DATE', 'PRCP']],
              on='DATE', how='outer')
df.head()
```

Each DataFrame had a column named `PRCP`, so the default is to add suffixes _x and _y to differentiate the columns. Normally, I would rename the columns at this point to something more informative. However, I know that I want to convert the DataFrame to a tidy format and I will modify the names later.

### Create a tidy data frame with columns for city and precipitation

```python
df
```

```python
df.describe()
```

How did this change the DataFrame?

`melt` turned the wide table (one precipitation column per city) into a long, tidy table: each row is now one observation — one city on one date — so the number of rows doubled and the city is stored as a value in the `CITY` column.

```python
# use DATE as the identifier variable
df = pd.melt(df, id_vars='DATE', var_name='CITY', value_name='precipitation')
```

### Rename columns or values to follow best practices

Rename the city values 'PWM' and 'SEA'

```python
df['CITY'] = df['CITY'].replace({'PRCP_x': 'SEA', 'PRCP_y': 'PWM'})
df['CITY'].unique()
```

##### Rename the columns to be lowercase using `df.rename()`

```python
df = df.rename(columns={'DATE': 'date', 'CITY': 'city', 'PRCP': 'precipitation'})
df.head()
```

##### The Portland, Maine data set does not have any `NaN` values of `precipitation`. Are any dates omitted?

**How many data points should we have from 2018 to 2022?**

```python
all_dates = pd.date_range('2018-01-01', '2022-12-31', freq='D')
print('Expected days per city:', len(all_dates))
df.groupby('city').size()
```

```python
# Make sure every city has a row for every date; dates missing from the file become NaN
full_index = pd.MultiIndex.from_product([all_dates, ['SEA', 'PWM']], names=['date', 'city'])
df = df.set_index(['date', 'city']).reindex(full_index).reset_index()

print(df.groupby('city').size())
print(df.groupby('city')['precipitation'].apply(lambda s: s.isna().sum()))
```

### Impute missing values

We will replace missing values with the mean across years of values on that day.


**Design an algorithm for replacing missing values with the mean across years of values on that day.**

**Algorithm**
1. Add columns for the month and day of each date.
2. For each city, group the rows by (month, day) and compute the mean precipitation across the years for that calendar day (`NaN` values are ignored).
3. Replace each missing precipitation value with the mean for its city and calendar day.
4. If a calendar day has no values in any year, fall back to the city's overall mean.

```python
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day

daily_mean = df.groupby(['city', 'month', 'day'])['precipitation'].transform('mean')
df['precipitation'] = df['precipitation'].fillna(daily_mean)

# fallback for any calendar day with no data in any year
city_mean = df.groupby('city')['precipitation'].transform('mean')
df['precipitation'] = df['precipitation'].fillna(city_mean)

df = df.drop(columns=['month', 'day'])

df.isna().sum()
```
